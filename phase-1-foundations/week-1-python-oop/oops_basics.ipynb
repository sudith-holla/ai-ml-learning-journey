```python
#throws error if we leave it as it is. if we wanna just create a class without any attributes or methods we can use "pass"
#class is basically a blueprint using which we create instances. It increases reusability and keeps the code simple and it should be used for repititive task,
#for example, if we want to maintain record for each employee in a company, we can use class with attributes and methods storing their information.
#by doing this, we dont have to use 100s of variable to store the data of each employee.
# class can have multiple instances. think of class as a big group and instances like members of the group.
```

```python
class Employee:
    raise_amount = 1.04
    no_of_emps = 0
    def __init__(self, name, age, salary): # init is a method, it automatically gets called when we initialize an instance.
        self.name = name                   # it takes arguments that we need to pass when we create the instance.
        self.age = age                     #self refers to the particular instance we r creating
        self.salary = salary
        Employee.no_of_emps += 1    #here im accessing the class variable using the class itself
                                    #because there is no use case of accessing it using instance
                                    #as having different values for no_of_employees for different instances makes no sense.

    def email(self):
      return '{}@company.com'.format(self.name)
# we can also create methods in a class if we want to get a task done and if we want to take an anction
#methods are nothing but functions inside a class.
    def apply_raise(self):
      self.salary=int(self.salary*self.raise_amount) # the only ways to access class variables are through class itslef or through instances.
      # if we access using instance, even though its not an attribute of than instance, python will look for the value in the class.
      #this allows us to change the classs variable for a particular instance.
# we have attributes whose values can change for each instance. Then we have
#class variables whose value remain the same for each nd every instance.
#we can change the value of class variable for a particular instance based on how its called in a method, but usually the same
#value is shared by all the instances.
# its initialized before init function outisde of methods.
    @classmethod # we can use class methods to globally change class variables as well
    def self_raise_amount(cls,amount):
      cls.raise_amount=amount
```

```python
Employee.self_raise_amount(1.05)
emp1=Employee("Sudith",22,100000)
emp2=Employee("Rhuthvik",25,80000)
```

#video 3 and 4 (Class methods, static methods and inheritance)

```python
# there are three types of methods in OOPS
#class methods automatically takes class as one of the arguments
#nomral methods automatically takes the instance as one of the arguments
#static methods dont take both class or instance as agruments

#lets understand the use case of class methods.
#they are used as alternative consturctors
#sometimes the data to create an instance maynot be in form the default constructor expects it to be
#so we use these class methods to bring the data to normal form and then create an instance of that particular class

class Student:
  def __init__(self,first,last,grade):
     self.first = first
     self.last= last
     self.grade=grade

  @classmethod
  def from_string(cls,data_string):
     first,last,grade=data_string.split('-')
     return cls(first,last,int(grade))

  @staticmethod
  def is_workday(day):
    if day.weekday()==5 or day.weekday()==6:
      return False
    return True

```

```python
#lets say we have data of 2 students as strings seperated by hyphens
stu1='Sudith-Holla-12'
stu2='Rhuthvik-Holla-18'

#inorder to create an instance using this data of the class student, we need
#to make sure that we have first,last and the grades stored somewhere
#and not in a single string. this is where class methods come in
#we can perform the necessary operations on the string using
#class methods to get the data we need.

stud1=Student.from_string(stu1)
stud2=Student.from_string(stu2)

#then we have static methods.
#we use this when we dont rely on both instabce and class. maybe
#to check whether a date is a working day or holiday for students
import datetime
my_date=datetime.date(2016,7,10)
print(Student.is_workday(my_date))

```

```python
#inheritance.Its very useful when a class(child) has same attributes
#and methods as another class(parent) along with its own attributes and methods
#this helps us improve readability and increase reusability
#keeps the code neat and clean
#for example, managers and developers can be subclasses of employee class
class Developer(Employee):
  raise_amount=1.10 #if we wanna change the value of the class variable of the parent class, we can just do it like this

  #lets say we wanna make some changes in the constructor of the parent class
  #lets say developer class takes ana extra argument.
  #in this case we dont have to rewrite the entire function, instead we can use super() to inherit the
  #exisiting constructor and then make changes to it.
  def __init__(self,name,age,salary,prog_lang):
    super().__init__(name,age,salary)
    self.prog_lang=prog_lang

  #so basically when we want to customize
  #a method in our subclass thats in the parent class and
  #define it again in the subclass, and when we call it,
  #python will priorotize the function defined in the subclass and
  #not the one in the parent class. So, we need super( ) to
  #get the data from parent class and modify it accordingly.
  #or we will have to rewrite the method again in subclass
  #and not use inheritance to our advantage

  def email(self):
     email=super().email()
     return f'{self.prog_lang}.{email}'



```

```python
dev1=Developer('Sudith',22,100000,'Python')
dev2=Developer('Rhuthvik',25,80000,'Java')
print(dev1.prog_lang)
dev1.apply_raise()
print(dev1.salary)
print(dev1.email())

```

```python
class Manager(Employee):
  def __init__(self,name,age,salary,employees=None):
    super().__init__(name,age,salary)
    if employees is None:
      self.employees=[]
    else:
      self.employees=employees

  def add_emp(self,emp):
    if emp not in self.employees:
      self.employees.append(emp)

  def remove_emp(self,emp):
    if emp in self.employees:
      self.employees.remove(emp)

  def print_emps(self):
    for emp in self.employees:
      print('-->',emp.name)
```
