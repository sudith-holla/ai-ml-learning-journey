```python
#throws error if we leave it as it is. if we wanna just create a class without any attributes or methods we can use "pass"
#class is basically a blueprint using which we create instances. It increases reusability and keeps the code simple and it should be used for repititive task,
#for example, if we want to maintain record for each employee in a company, we can use class with attributes and methods storing their information.
#by doing this, we dont have to use 100s of variable to store the data of each employee.
# class can have multiple instances. think of class as a big group and instances like members of the group.
```

```python
class Employee:
    raise_amount = 1.04
    no_of_emps = 0
    def __init__(self, name, age, salary): # init is a method, it automatically gets called when we initialize an instance.
        self.name = name                   # it takes arguments that we need to pass when we create the instance.
        self.age = age                     #self refers to the particular instance we r creating
        self.salary = salary
        Employee.no_of_emps += 1    #here im accessing the class variable using the class itself
                                    #because there is no use case of accessing it using instance
                                    #as having different values for no_of_employees for different instances makes no sense.

    def email(self):
      return '{}@company.com'.format(self.name)
# we can also create methods in a class if we want to get a task done and if we want to take an anction
#methods are nothing but functions inside a class.
    def apply_raise(self):
      self.salary=int(self.salary*self.raise_amount) # the only ways to access class variables are through class itslef or through instances.
      # if we access using instance, even though its not an attribute of than instance, python will look for the value in the class.
      #this allows us to change the classs variable for a particular instance.
# we have attributes whose values can change for each instance. Then we have
#class variables whose value remain the same for each nd every instance.
#we can change the value of class variable for a particular instance based on how its called in a method, but usually the same
#value is shared by all the instances.
# its initialized before init function outisde of methods.
```

```python
emp1=Employee("Sudith",22,100000)
emp2=Employee("Rhuthvik",25,80000)
```
