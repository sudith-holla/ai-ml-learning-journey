````python
array_6=np.array([[1,2,0,0,4,0],[2,3,4,5,6,7]])
indices=np.where(array_6!=0) #np.where to find the indices of values in an array that satisfy a condition
print(indices) #np.where returns tupes of arrays, 1st array has the row indices and 2nd array has column indices
print(np.transpose(indices)) #to get the indices in the form of list of tuples, transpose the tuple of arrays


````
